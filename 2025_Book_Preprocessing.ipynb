1:3

```python
# 베스트 셀러 장르 비율
genre_distribution = {
    '소설/시/희곡': 60.065133,
    '에세이': 14.438399,
    '자기계발': 10.178680,
    '사회 정치': 15.317788
}

# 목표 총 개수
target_count = 4887

# 각 장르별 샘플 수 계산
samples_per_genre = {genre: int((percentage / 100) * target_count) for genre, percentage in genre_distribution.items()}

# 샘플 수가 해당 장르의 데이터 개수를 초과하지 않도록 조정
downsampled_non_bestsellers = pd.DataFrame()

for genre, num_samples in samples_per_genre.items():
    genre_data = non_bestsellers[non_bestsellers['장르'] == genre]
    if len(genre_data) < num_samples:
        num_samples = len(genre_data)  # 장르 데이터가 부족하면 전체 데이터 개수로 설정
    genre_samples = genre_data.sample(n=num_samples, replace=True)  # 필요시 중복 허용
    downsampled_non_bestsellers = pd.concat([downsampled_non_bestsellers, genre_samples])

# 최종 데이터프레임 생성
final_dataset_3 = pd.concat([bestsellers, downsampled_non_bestsellers])

# 결과 출력 및 확인
print("총 데이터 개수:", len(final_dataset_3))
final_dataset_3
```

```python
# 장르별 데이터 개수
genre_counts = final_dataset_3['장르'].value_counts()

# 장르별 비율 계산 (전체 대비 비율)
genre_percentage = genre_counts / len(final_dataset_3) * 100

# 결과 출력
genre_percentage
```
