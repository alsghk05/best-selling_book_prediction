### 1. 데이터 불러오기

```python
import pandas as pd
final_dataset_3 = pd.read_csv('book_1_3.csv')
final_dataset_3
```

데이터 셔플(Data Shuffle)

```python
import random
df_dataset_3_shuffled = final_dataset_3.sample(frac=1, random_state=random.randint(0, 10000)).reset_index(drop=True)
df_dataset_3_shuffled
```

### 2. Embedding

- kobert

```python
from kobert_transformers import get_tokenizer, get_kobert_model
import torch
import pandas as pd
import numpy as np

# KoBERT 모델 및 토크나이저 불러오기
model = get_kobert_model()
tokenizer = get_tokenizer()

# 임베딩을 위한 함수 (KoBERT로 임베딩, 각 문장을 벡터로 변환)
def get_kobert_embeddings(text_list, max_length=128):
    embeddings = []
    
    for text in text_list:
        # None 및 NaN 값을 빈 문자열로 처리하고, 그 외 값은 문자열로 변환
        if pd.isna(text):
            text = ""  # None 또는 NaN은 빈 문자열로 대체
        else:
            text = str(text)  # 그 외는 문자열로 변환
        
        inputs = tokenizer(text, return_tensors='pt', padding=True, truncation=True, max_length=max_length)
        with torch.no_grad():
            outputs = model(**inputs)
        embedding = outputs.last_hidden_state.mean(dim=1).squeeze().numpy()
        embeddings.append(embedding)
    
    return np.array(embeddings)


# 도서 이름, 줄거리, 작가, 출판사, 장르 각각에 대해 임베딩 수행
book_titles = df_dataset_3_shuffled['도서 이름'].tolist()
summaries = df_dataset_3_shuffled['줄거리'].tolist()
authors = df_dataset_3_shuffled['작가'].tolist()
publishers = df_dataset_3_shuffled['출판사'].tolist()
genres = df_dataset_3_shuffled['장르'].tolist()

# 각 열에 대해 KoBERT 임베딩
book_title_embeddings = get_kobert_embeddings(book_titles)
summary_embeddings = get_kobert_embeddings(summaries)
author_embeddings = get_kobert_embeddings(authors)
publisher_embeddings = get_kobert_embeddings(publishers)
genre_embeddings = get_kobert_embeddings(genres)

# 임베딩 결합: 도서 이름, 줄거리, 작가, 출판사, 장르
final_embeddings_3 = np.concatenate(
    (book_title_embeddings, summary_embeddings, author_embeddings, publisher_embeddings, genre_embeddings), axis=1
)

# 베스트셀러 여부 라벨 설정 (Y -> 1, N -> 0)
labels_3 = df_dataset_3_shuffled['best'].apply(lambda x: 1 if x == 'Y' else 0).values
```

word2vec+one hot encoding

```python
from konlpy.tag import Okt
from gensim.models import Word2Vec
import pandas as pd
import numpy as np

# Okt 객체 생성
okt = Okt()

# 텍스트 토큰화 함수
def tokenize_text(text_list):
    tokenized_texts = []
    for text in text_list:
        if pd.isna(text):
            text = ""  # None 또는 NaN은 빈 문자열로 대체
        else:
            text = str(text)  # 그 외는 문자열로 변환
        tokens = okt.morphs(text)  # Okt 형태소 분석기로 토큰화
        tokenized_texts.append(tokens)
    return tokenized_texts

# Word2Vec 임베딩 학습 함수
def train_word2vec(tokenized_texts, vector_size=100, window=5, min_count=1, workers=4):
    model = Word2Vec(sentences=tokenized_texts, vector_size=vector_size, window=window, min_count=min_count, workers=workers)
    return model

# 텍스트 리스트를 Word2Vec 임베딩 벡터로 변환하는 함수
def get_word2vec_embeddings(text_list, model, vector_size=100):
    embeddings = []
    for tokens in text_list:
        if len(tokens) == 0:
            embeddings.append(np.zeros(vector_size))  # 토큰이 없는 경우 0으로 채움
        else:
            embedding = np.mean([model.wv[token] for token in tokens if token in model.wv], axis=0)
            if np.isnan(embedding).any():
                embedding = np.zeros(vector_size)  # 모든 토큰이 모델에 없을 경우 0으로 채움
            embeddings.append(embedding)
    return np.array(embeddings)

# 데이터셋의 텍스트 열
book_titles = df_dataset_3_shuffled['도서 이름'].tolist()
summaries = df_dataset_3_shuffled['줄거리'].tolist()
authors = df_dataset_3_shuffled['작가'].tolist()

# 텍스트 토큰화
book_title_tokens = tokenize_text(book_titles)
summary_tokens = tokenize_text(summaries)
author_tokens = tokenize_text(authors)

# Word2Vec 모델 학습 (모든 텍스트를 합쳐서 학습)
total_tokens = book_title_tokens + summary_tokens + author_tokens
word2vec_model = train_word2vec(total_tokens)

# 각 열에 대해 Word2Vec 임베딩
book_title_embeddings = get_word2vec_embeddings(book_title_tokens, word2vec_model)
summary_embeddings = get_word2vec_embeddings(summary_tokens, word2vec_model)
author_embeddings = get_word2vec_embeddings(author_tokens, word2vec_model)

# 출판사 및 장르에 대해 원핫 인코딩 수행
publishers_onehot = pd.get_dummies(df_dataset_3_shuffled['출판사'], prefix='출판사').values
genres_onehot = pd.get_dummies(df_dataset_3_shuffled['장르'], prefix='장르').values

# 임베딩 결합: 도서 이름, 줄거리, 작가, 출판사(원핫), 장르(원핫)
final_embeddings_333 = np.concatenate(
    (book_title_embeddings, summary_embeddings, author_embeddings, publishers_onehot, genres_onehot), axis=1
)

# 베스트셀러 여부 라벨 설정 (Y -> 1, N -> 0)
labels_333 = df_dataset_3_shuffled['best'].apply(lambda x: 1 if x == 'Y' else 0).values
```

##### Train-Test Split

```python
import numpy as np
# 전체 라벨 배열이 'labels'라고 가정합니다.
# 베스트셀러(1)의 비율 계산
best_seller_ratio = np.mean(labels_3)

# 비베스트셀러(0)의 비율 계산
non_best_seller_ratio = 1 - best_seller_ratio

# 결과 출력
print(f"전체 데이터에서 베스트셀러의 비율: {best_seller_ratio:.2f}")
print(f"전체 데이터에서 비베스트셀러의 비율: {non_best_seller_ratio:.2f}")

# 베스트셀러와 비베스트셀러의 개수 출력
best_seller_count = np.sum(labels_3)
non_best_seller_count = len(labels_3) - best_seller_count

print(f"베스트셀러 개수: {best_seller_count}")
print(f"비베스트셀러 개수: {non_best_seller_count}")
print(f"전체 도서 수: {len(labels_3)}")
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
import numpy as np

# KoBERT 임베딩 및 라벨 가져오기 (이미 처리된 데이터라고 가정)
X = final_embeddings_3  # 도서 정보 임베딩
y = labels_3  # 베스트셀러 여부 (1: 베스트셀러, 0: 비베스트셀러)

# StratifiedKFold 설정
k = 5  # 5겹 교차 검증
skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)

# K-Fold를 통해 Train/Validation으로 분리된 데이터셋 비율 유지
for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    # 훈련 데이터 및 검증 데이터 분리
    X_train_fold, X_val_fold = X[train_idx], X[val_idx]
    y_train_fold, y_val_fold = y[train_idx], y[val_idx]

    # 각 폴드의 베스트셀러 여부 비율 확인
    best_train_ratio = np.mean(y_train_fold)
    best_val_ratio = np.mean(y_val_fold)

    print(f"Fold {fold + 1}:")
    print(f"  Train set - Best seller ratio: {best_train_ratio:.2f}")
    print(f"  Validation set - Best seller ratio: {best_val_ratio:.2f}")
```
